Ethical issues regarding data collection may be divided into the following categories:
- Consent: Individuals must be informed and give their consent for information to be collected.
- Ownership: Anyone collecting data must be aware that individuals have ownership over their information.
- Intention: Individuals must be informed about what information will be taken, how it will be stored, and how it will be used.

### Tree Census Example ###

Congratulations! You’ve just been hired as a census taker for the North American tree census! This census will be a partnership between Canada and the United States. You’ve agreed to use the Imperial measurements (the American system), so will be collecting the data in feet.
You’ve been assigned to the North Atlantic Region. That includes both the Northeast United States and Eastern Canada.

‘Height’ and ‘Species’ are our variables. Each individual tree is called an entity, observation, or instance. In a well-organized dataset, the variables describe a characteristic of our entities. However, it can be surprisingly difficult to define good variables. Good variables measure only one characteristic and should not be a characteristic themselves. 

We are looking at trees along city streets, highways, and in undeveloped areas. We also want to know if trees are standing alone or with others.

In our tree census, we are collecting data about two types of variables: one that we measure (height) and one that we categorize (species).
The difference between measuring and categorizing is so important that the data itself is termed differently:
- Variables that are measured are Numerical variables.
  - Numerical variables are a combination of the measurement and the unit.
  - There are two ways to get a number: by counting and measuring. Counting gives us whole numbers and discrete variables. Measuring gives us potentially partial values and continuous variables
- Variables that are categorized are Categorical variables.
  - Categorical variables describe characteristics with words or relative values.
    - nominal variable which literally means a named value
    - Dichotomous variables have only 2 logical possibilities
    - Ordinal variables are inherently ordered. “on a scale of 1-5 how much do you agree with this statement?” This is called a likert scale. can get a little confusing because they are often represented as numbers. But they don’t represent measurements or counts, they represent categories

Messy Data Problems
Different problems need to be handled differently, so let’s categorize them:
Typos like Tuuullip for Tulip
Missing data like the Pin Oak (tree ID 11222) that doesn’t have a height
Inconsistent coding like the Pin Oak (tree 18564)’s Prettiness value is ‘three’ rather than ‘3’ and the Single value for all of our trees is ‘no’ rather than ‘0’.

If we don’t fix these issues, we will likely end up with problems in our analysis. For example:
Tulip trees might be divided into 2 categories
We might get an inaccurate average height for Pin Oaks because we are missing a data point
Our computer might return an error message when we try to group trees into their Prettiness value or find all of the trees that grow alone.

Missing Data
- missing completely at random: no deeper meaning as to why the data is missing, it just wasn't entered properly (hungry and tired)
- missing at random: we can predict if one value is missing based on the value in another variable (tape measure was too short)
- structurally missing: a value wasn't expected to begin with (no visible fruit on the tree, so we can't count it)

Accuracy
- thinking about data against expectations and common sense is crucial for spotting issues with accuracy. you can do this by inspecting the distribution and outliers to get clues about what the data looks like.
- critically considering how error could have crept in during the data collection process will help you group and evaluate the data to uncover systematic inconsistencies.
- identifying ways that duplicate values could have been created goes a long way towards ensuring that reality is only represented once in your data. a useful technique is to distinguish between what was human collected versus programmatically generated and using that distinction to segment the data.

Validity is a special kind of quality measure because it’s not just about the dataset, it’s about the relationship between the dataset and its purpose. A dataset can be valid for one question and invalid for another.
The best way to spot issues with the validity of a dataset is to ask: Does this variable measure what I think it does?

Representative Samples
- goal: to find a mix of observations that contains all of the features in the larger population
- convenience sample: sampling only what's convenient, not good for representing a broader population
- bias is introduced by constraining the sample
- best practice is to create a sample that represents the entire population
- population: all of the trees in the North Atlantic region
- sample: the trees that we have data about

### Statistical Thinking ###

- summary statistics: used to measure, describe, communicate, and understand more specific information about variables in a dataset
- to start a summary, describe some of the categorical variables such as frequency (the count of something), proportion (frequency divided by the total), percentage (the proportion converted from a decimal to a percentage)
- next, explore numeric/qualitative data
- distribution: a function that shows all possible values of a variable and how frequently each value occurs
- bell/hill shaped: normal distribution
  
Mean and Standard Deviation
- mean: average, describes the center of a numeric distribution by adding all values and dividing by the count
- standard deviation: the spread of values in a numeric distribution relative to the mean. calculated by finding the average squared distance from each data point to the mean and square-rooting the result. more influenced by outliers and skewness than the IQR

Skewed Distributions
- asymmetrical with a steep change in frequency on one side and a flatter, trailing change on the other
- right-skewed/positively-skewed: tail on the right, the higher values in the tail will pull the mean out to be higher than the median
- left-skewed/negatively-skewed: tail on the left, the lower values in the tail will pull the mean out to be lower than the median

Median and IQR
- median: the middle value when all values are arranged from smallest to largest, 50th percentile or second quartile (Q2)
- interquartile range (IQR): a quartile is a marker for a quarter (25%) of the data (Q1 = first quartile, Q2 = second quartile (median), Q3 = third quartile). the IQR is the difference between Q3 and Q1, marking the range for the middle 50% of data. 

Outliers and Robust Measures
- outliers: extreme values that are distant from the rest of the distribution, tend to influence the mean more heavily than the median
- robust statistics: not heavily influenced by extreme values (IQR and median)

Aggregate Data
- separation and summarization of data, a way to explore relationships between variables
- mode: the value with the highest frequency, where the peak of the distribution occurs
- bimodal distribution: has two modes

Variable Relationships
- correlation coefficient: this number ranges from -1 to +1 and tells us two things about linear relationships
  - direction: a positive coefficient means that higher values in one variable are associated with higher values in the other, a negative coefficient means higher values in one variable are associated with lower values of the other
  - strength: the farther the coefficient is from 0, the stronger the relationship and the more points in a scatter plot look like a line

### Visualizing Data ###
translating the data relationship to a visual relationship

1. We visualize data to accomplish a goal – to explore and see something new in the data, to strengthen an argument with a visual aid, to share a data-driven idea with someone else, and more
2. Our design decisions can either help or hinder how effectively a data viz communicates. We tell stories with data viz using good design choices in chart type, color, labels, context, and more.

Different types of charts
- line chart or bar chart: shows a change over time
- pie chart: shows parts of a whole
- histogram: shows the spread of data points in one variable (distribution)
- scatterplot: shows a direct comparison of two numeric variables to help understand a trend
  
- univariate: shows a change in one variable, measures how much, count or distribution (bar graph, histogram, density curve, box plot, violin plot, univariate map)
- bivariate and multivariate: shows the relationships between two or more variables (scatterplot, line chart, bivariate map)

Aesthetic Properties
attributes we use to communicate data visually
- position
- size
- shape
- color/pattern

information redundancy: encoding the same information in different visual properties, helpful for prioritizing values, helps key data points stand out, ensures that colorblind viewers can access all the information

consider the audience: provide necessary details, include context that's helpful for the specific audience, avoid excess graphics, annotations, and general lines that don't actually contain information

Accessibility Basics
- colorblind-friendly color palettes
- large enough font size
- readable, web-accessible font type
- alt text on data visualization images online

Universal Design
- readability: keep the reading level to a high school level whenever possible
- prior knowledge: define unfamiliar terms and avoid unnecessary jargon
- information overload: introduce new information with intentional pacing and organization

Misleading and Confusing Graphs
- x-axis (horizontal)
- y-axis (vertical)
- generally, we measure time horizontally
- break: a break starts the count at a number that is not zero
  - keep enough context to view differences in proportion to a meaningful amount, OR
  - make two graphs, one without a break and one "zoomed in," OR
  - choose a visualization type that shows the change, rather than the raw numbers
  
- scaling: refers to the distances between numbers on an axis, almost all graphs use a linear scale (numbers count up by a consistent interval)
  - logarithmic scale: log scale, shows exponential growth that won't fit on the page with a linear scale, not a good choice for a general audience
  
- color scales: 
  - sequential scales: colors in a sequence, same hue with more and more white added or taken away, used to show a variable increasing or decreasing in intensity or amount
  - divergent scales: anchored by colors from opposite sides of the color wheel, complementary colors, used where the middle is a baseline and either side represents a contrasting change, do a good job of showing a positive/negative swing
  - categorical scales: use a variety of colors to differentiate categories without assigning a rank or order to them
- color associations: we tend to view darker colors as "more" and lighter colors as "less"
  - pre-existing associations for certain colors: culturally specific or influenced by the norms of a particular field
    - sometimes it's best to stick with norms (red for republicans, blue for democrats) but not always (pink for women, blue for men)

- labels and titles: helps the viewer understand what is going on, call out points of interest, explain outliers, include background information

### Analyzing Data ###

- data analysis: discovering useful information, informing conclusions, and supporting decision-making. evaluates the presence, strength, and direction of relationships between different variables in our data.
  - the five types of data analysis
    - descriptive analysis: lets us describe, summarize, and visualize data so that patterns can emerge. include measures of central tendency (range, quartiles, variance, standard deviation, distribution) which are referred to as descriptives or summary statistics. lets us draw basic conclusions about the data at hand. cannot extend findings to other datasets
    - exploratory analysis: shows us underlying patterns and relationships within datasets. cannot determine causation
      - clustering: unsupervised machine learning. these techniques "learn" patterns from untagged data, and help us see relationships between many variables at once
      - principal component analysis (PCA) - compresses the variables into principal components that can be plotted against each other, we can then use k-means clustering to look for trends
    - inferential analysis: lets us test a hypothesis on a sample of a population and then extend our conclusions to the whole population (a/b testing)
      - sample size must be big enough compared to the total population size (10%)
      - our sample must be randomly selected and representative of the total population
      - we can only test one hypothesis at a time
    - causal analysis: generally relies on carefully design experiments
      - only change one variable at a time
      - carefully control all other variables
      - are repeated multiple times with the same results
      - replication, randomization, and control are key components
      - correlation does not equal causation
    - predictive analysis: uses data and supervised machine learning to identify the likelihood of future outcomes (regression models, support vector machines, deep learning convolutional neural networks). garbage in, garbage out...a predictive model trained on poor-quality data will make poor-quality predictions
      - supervised machine learning algorithms are trained with labeled data and predict the likelihood of future outcomes
      - supervised machine learning algorithms can only be as good as the data used to train them

### Bias in Data Analysis ###
- automation bias: stems from the idea that computers or machines are more trustworthy than humans because they are more objective
- selection bias/sample bias: occurs when study subjects are not representative of the population
- historical bias: systematic influence based on historic social and cultural beliefs
- algorithmic bias: arises when an algorithm produces systematic and repeatable errors that lead to unfair outcomes, such as privileging one group over another
- evaluation bias: testing an algorithm with a non-representative dataset
- confirmation bias: our tendency to seek out information that supports our views
- overgeneralization bias: inappropriately extending observations make with one dataset to other datasets
- reporting bias: the human tendency to only report or share results that affirm our beliefs or hypothesis

```python
CREATE TABLE celebs (
   id INTEGER, 
   name TEXT, 
   age INTEGER
);
```

- CREATE TABLE is a clause that tells SQL you want to create a new table
- celebs is the name of the table
- id INTEGER, name TEXT, age INTEGER is a list of parameters defining each column or attribute and its data type

```python
INSERT INTO celebs (id, name, age) 
VALUES (1, 'Justin Bieber', 29);
```

- the INSERT statement inserts a new row into a table. use this to add new records
- VALUES is a clause that indicates the data being inserted
- 1, 'Justin Bieber', 29 is a parameter identifying the values

SELECT statements are used to fetch data from the database. the statement below selects all data in the 'name' column

```python
SELECT name FROM celebs;
```

- SELECT is a clause that indicates a statement is a query
- name specifies the column to query the data from
- FROM celebs specifies the name of the table
- you can query all columns using SELECT *. this is a wildcard character
- SELECT statements always return a new table called the result set

- ALTER TABLE adds a new column to the table. the statement below adds a new column twitter_handle to the celebs table
- NULL is a special value that represents missing or unknown data. the rows that existed before the column was added will have NULL values for twitter_handle

```python
ALTER TABLE celebs 
ADD COLUMN twitter_handle TEXT;
```

- UPDATE edits a row in a table. use this when you want to change existing records
- the statement below updates the record id 4 to have the twitter_handle of @taylorswift13

```python
UPDATE celebs 
SET twitter_handle = '@taylorswift13' 
WHERE id = 4; 
```

- DELETE FROM deletes one or more rows
- the statement below deletes all records in the celebs table with no twitter_handle

```python
DELETE FROM celebs 
WHERE twitter_handle IS NULL;
```

constraints that add information about how a column can be used are invoked after specifying the data type for a column. used to tell the database to reject inserted data that does not adhere to a certain restriction.

```python
CREATE TABLE celebs (
   id INTEGER PRIMARY KEY, 
   name TEXT UNIQUE,
   date_of_birth TEXT NOT NULL,
   date_of_death TEXT DEFAULT 'Not Applicable'
);
```

- PRIMARY KEY columns can be used to uniquely identify a row. attempts to insert a row with an identical value to a row already in the table will result in a constraint violation which will not allow the row to be inserted
- UNIQUE columns have different values for every row. similar to primary key except the table can have many different unique columns
- NOT NULL columns must have a value. attempts to insert a row without a value for a not null column will result in a constraint violation and will not be inserted
- DEFAULT columns take an additional argument that will be the assumed value for an inserted row if the new row does not specify a value for that column
  
- we can select individual columns by their names, separated by a comma

```python
SELECT column1, column2
FROM table_name;
```

- AS is a keyword that allows you to rename a column or table using an alias
- when using AS, the columns are not being renamed in the table, it only appears in the result

```python
SELECT name AS 'Titles'
FROM movies;
```

- DISTINCT is used to return unique values in the output. it filters out all duplicate values in the specified column

```python
SELECT DISTINCT tools
FROM inventory;
```

- we can restrict our query results using the WHERE clause
- this only includes rows where the following condition is true
- comparison operators:
  - = equal to
  - != not equal to
  - > greater than
  - < less than
  - >= greater than or equal to
  - <= less than or equal to

```python
SELECT *
FROM movies
WHERE imdb_rating > 8;
```

- LIKE is a special operator used with the WHERE clause to search for a specific pattern in a column

```python
SELECT *
FROM movies
WHERE name LIKE 'Se_en';
```

- name LIKE 'Se_en' is a condition evaluating the name column for a specific pattern
- Se_en represents a pattern with a wildcard character
- the _ means you can substitute any character here
- the % matches zero or more missing characters in the pattern

```python
SELECT *
FROM movies
WHERE name LIKE 'A%';
```

- A% matches all movies that begin with the letter A
- %a would match all movies that end with the letter a
- we can also use % before and after a pattern

```python
SELECT *
FROM movies
WHERE name LIKE '%man%';
```

- any movie that contains the word 'man' in its name would be returned in the result
- LIKE is not case sensitive

- unknown values are indicated by NULL
- it is not possible to test for NULL values with comparison operators, we have to use IS NULL and IS NOT NULL
- to filter all movies with an imdb rating:

```python
SELECT name
FROM movies
WHERE imdb_rating IS NOT NULL;
```

- the BETWEEN operator is used in a WHERE clause to filter the result set within a certain range
- it accepts two values that are either numbers, text, or dates
- this statement filters the result set to only include movies with years from 1990 up to and including 1999:

```python
SELECT *
FROM movies
WHERE year BETWEEN 1990 AND 1999;
```

- when the values are text, BETWEEN filters the result set for within the alphabetical range (up to but not including)
